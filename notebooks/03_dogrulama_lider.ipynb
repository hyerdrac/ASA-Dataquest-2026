```python
import pandas as pd

veri = pd.read_csv("../data/clean/countries_clean.csv")
v = veri[veri.Time <= 2100]   # 2101'de yalnızca 1 Ocak nüfusu var

# Kontrol #1: Dünya nüfusunun zirvesi (nüfus bin kişi cinsinden; 1e6'ya bölünce milyar)
dunya = v.groupby("Time").TPopulation1July.sum() / 1e6
print("Zirve yılı:", dunya.idxmax(), "| Zirve:", round(dunya.max(), 2), "milyar")
```

```python
# Kontrol #2: Bölgelerin dünya nüfusundaki payı (%)
pay = v.pivot_table(index="Time", columns="Region", values="TPopulation1July", aggfunc="sum")
pay = pay.div(pay.sum(axis=1), axis=0) * 100
pay.loc[[1950, 2024, 2100], ["Africa", "Europe"]].round(1)
```
